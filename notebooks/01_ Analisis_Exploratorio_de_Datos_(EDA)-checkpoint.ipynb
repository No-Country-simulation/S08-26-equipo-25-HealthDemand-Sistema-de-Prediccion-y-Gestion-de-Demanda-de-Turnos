# Fase 1: Análisis Exploratorio de Datos (EDA)

El propósito de este notebook es auditar y analizar el historial transaccional de turnos médicos correspondientes al período 2023-2025. Antes de iniciar el modelado predictivo, es fundamental transformar los datos crudos en conocimiento. 

A lo largo de este análisis, exploraremos cómo interactúan diversas variables operativas (como la sede, la especialidad y el día de la semana) y conductuales (como la anticipación de la reserva) con el resultado final del turno (Atendido, Cancelado o Ausente). El objetivo es identificar visual y estadísticamente los cuellos de botella y los detonantes del ausentismo (*No-Show*) para justificar las decisiones de negocio y fundamentar la posterior ingeniería de características (*Feature Engineering*).

#### Importamos librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### 1. Carga de datos

```python
df = pd.read_csv('healthdemand_historico_23_25.csv', parse_dates=['fecha_solicitud', 'fecha_hora_turno'])
```

```python
# Configuración visual
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (14, 6)
```

#### 2. Análisis 1: Mapa de Calor de Demanda (Día vs Hora)

```python
df['dia_semana'] = df['fecha_hora_turno'].dt.day_name()
df['hora_dia'] = df['fecha_hora_turno'].dt.hour
orden_dias = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

matriz_demanda = df.pivot_table(index='hora_dia', columns='dia_semana', aggfunc='size', fill_value=0)
matriz_demanda = matriz_demanda[orden_dias]

plt.figure(figsize=(12, 8))
sns.heatmap(matriz_demanda, cmap="YlOrRd", annot=True, fmt="d", linewidths=.5)
plt.title('Mapa de Calor: Volumen de Turnos por Día y Hora', fontsize=16)
plt.xlabel('Día de la Semana')
plt.ylabel('Hora del Día')
plt.show()
```

#### 3. Análisis 2: Tasa de Ausentismo por Especialidad

```python
tasa_ausentismo = df[df['estado_turno'] == 'Ausente'].groupby('especialidad').size() / df.groupby('especialidad').size()
tasa_ausentismo = tasa_ausentismo.sort_values(ascending=False) * 100

plt.figure(figsize=(10, 5))
sns.barplot(x=tasa_ausentismo.values, y=tasa_ausentismo.index, palette="Reds_r")
plt.title('Tasa de Ausentismo (No-Show) por Especialidad (%)', fontsize=16)
plt.xlabel('Porcentaje de Ausentismo (%)')
plt.ylabel('')
plt.show()
```

#### 4. Análisis 3: Impacto del "Lead Time" en la Asistencia

```python
df['lead_time'] = (df['fecha_hora_turno'] - df['fecha_solicitud']).dt.days
df['lead_time_bin'] = pd.cut(df['lead_time'], bins=[-1, 7, 14, 30, 60], labels=['0-7 días', '8-14 días', '15-30 días', '31-60 días'])

ausentismo_lead = df[df['estado_turno'] == 'Ausente'].groupby('lead_time_bin', observed=True).size() / df.groupby('lead_time_bin', observed=True).size() * 100

plt.figure(figsize=(8, 5))
sns.lineplot(x=ausentismo_lead.index, y=ausentismo_lead.values, marker='o', linewidth=2, color='darkred')
plt.title('Riesgo de Ausentismo según Anticipación del Turno (Lead Time)', fontsize=16)
plt.xlabel('Anticipación de la Reserva')
plt.ylabel('Probabilidad de Ausencia (%)')
plt.ylim(0, 30)
plt.show()
```

## Conclusiones y Hallazgos Estratégicos (Insights de Negocio)

Tras la exploración exhaustiva del dataset, hemos validado que el comportamiento de los pacientes sigue patrones medibles y predecibles, lo que confirma la viabilidad técnica del sistema **HealthDemand**. Los hallazgos más críticos para la operación son:

1. **El "Lead Time" como detonante de riesgo:** La distancia temporal entre la solicitud y la fecha del turno es el factor más influyente en el ausentismo. Se evidenció que los turnos solicitados con más de **30 días de anticipación** cuadruplican el riesgo de *No-Show* (alcanzando tasas cercanas al 25%) frente a los agendados en la misma semana (~7.5%). Este *insight* justifica técnica y financieramente la implementación de *overbooking* inteligente para citas a largo plazo.
2. **Disparidad Disciplinaria:** El comportamiento de asistencia está fuertemente segmentado por el tipo de atención médica. Mientras que **Psiquiatría** y **Pediatría** concentran los mayores índices de ausencias y cancelaciones históricas, los pacientes de **Oncología** presentan una fidelidad de asistencia superior al 88%. El futuro modelo de clasificación deberá ponderar de forma distinta cada disciplina.
3. **Puntos Ciegos en la Agenda (Efecto Fin de Semana):** Se detectó una concentración de cancelaciones de última hora los días viernes por la tarde, lo que actualmente genera subutilización de recursos y horas médicas ociosas que podrían reasignarse dinámicamente.
